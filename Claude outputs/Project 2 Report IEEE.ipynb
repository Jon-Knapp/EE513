```python
import json
from pathlib import Path
import base64
import pandas as pd
from IPython.display import HTML, display

# results2.json follows results2.template.json from the Canvas page ([value, uncertainty] pairs, null if not
# measured). project2_checks.json holds the tables below, written by the analysis scripts:
#   "constants":      {"saturation_dn": .., "wb_gains": [r, g, b], "ccm_cam_to_xyz": [[..]*3]*3}
#   "ablation":       [{"change": str, "psnr_db": [v, u], "delta_db": [v, u], "note": str}, ...]
#   "baseline_psnr_db": [v, u]
#   "denoise_eval":   [{"iso": int, "method": str, "psnr_db": [v, u], "ssim": [v, u], "lpips": [v, u]}, ...]
#   "latency":        [{"what": str, "size": str, "hardware": str, "ms": [v, u]}, ...]
#   "stage_times":    [{"stage": str, "ms": [v, u]}, ...]
#   "detector":       [{"version": str, "objects": int, "mean_conf": [v, u], "agreement": [v, u]}, ...]
#   "detector_info":  {"model": str, "threshold": float, "n_scenes": int, "agreement_metric": str}
def _load(name):
    p = Path(name)
    return json.loads(p.read_text(encoding="utf-8")) if p.exists() else {}

R = _load("results2.json")
C = _load("project2_checks.json")
P1 = _load(Path("..") / "Project 1" / "results.json")


def pmv(v, fmt="{:.3g}"):
    """Format a [value, uncertainty] pair; None or missing -> 'not measured'."""
    if v is None:
        return "not measured"
    if isinstance(v, (list, tuple)):
        return f"{fmt.format(v[0])} ± {fmt.format(v[1])}".replace("-", "−")
    return fmt.format(v)


def pm(key, fmt="{:.3g}", src=None):
    return pmv((src if src is not None else R).get(key), fmt)


def show(path, cap, width=None):
    if isinstance(path, (list, tuple)):
        missing = [p_ for p_ in path if not Path(p_).exists()]
        if missing:
            return pending(", ".join(missing), cap)
        from PIL import Image as _I
        ar = [_I.open(p_).width / _I.open(p_).height for p_ in path]
        W = float(str(width or "100%").rstrip("%"))
        imgs = "".join(f'<img src="data:image/png;base64,{base64.b64encode(Path(p_).read_bytes()).decode()}" '
                       f'style="width:{W * a_ / sum(ar) - 1:.1f}% !important; display:inline-block; margin:0 0.5%;">'
                       for p_, a_ in zip(path, ar))
        display(HTML(f'<div class="ieee-fig"><div style="text-align:center">{imgs}</div><div class="fig-caption">{cap}</div></div>'))
        return
    if not Path(path).exists():
        return pending(path, cap)
    b64 = base64.b64encode(Path(path).read_bytes()).decode()
    style = f' style="width:{width} !important"' if width else ""
    display(HTML(f'<div class="ieee-fig"><img src="data:image/png;base64,{b64}"{style}><div class="fig-caption">{cap}</div></div>'))


def pending(what, cap):
    display(HTML(f'<div class="ieee-fig"><div style="border:0.75pt dashed #888; padding:14pt; text-align:center; '
                 f'font-size:8pt; color:#666">[pending: {what}]</div><div class="fig-caption">{cap}</div></div>'))


def table(rows, columns, caption):
    if not rows:
        display(HTML(f'<div class="table-caption">{caption}</div>'
                     f'<p style="text-align:center; font-size:8pt; color:#666; text-indent:0">[pending]</p>'))
        return
    df = pd.DataFrame(rows, columns=columns)
    display(df.style.hide(axis="index").set_caption(caption))
```

## II. Ground Truth

### A. Denoising: The Averaged Burst

The bursts share one static scene, the cover of a book under a steady lamp, with the phone braced [on a tripod / on a solid support]: 32 frames at ISO 50 (1/15 s), 16 at ISO 200 (1/60 s) and 16 at ISO 2500 (1/125 s). The exposures keep the raw level about the same in all three, so the bursts differ in photon count and gain, not in brightness. I work on one set of green sites (half resolution), align every frame to the first by phase correlation on a Hann-windowed pair, and average. Two details of `cv2.phaseCorrelate` mattered: on the 2016-wide green plane it pads to a 2025-point DFT and reports a 0.5 px shift for an image against itself, so shifts are estimated on a 2000 × 1500 crop that needs no padding; and when it doesn't pad, it multiplies its inputs by the window in place. On frame 1 shifted by five known amounts up to 2.1 px, the estimator's error was at most 0.11 px (scikit-image's upsampled phase correlation, kept as a cross-check, reached 0.17 px). In the ISO 2500 burst the largest shift from frame 1 is 0.21 px and the median 0.13 px. One ISO 2500 frame's noise, the MAD of the difference of two aligned frames over 0.6745√2, is 0.0123 of full scale (41.9 DN), with the 1.7% of pixels near saturation left out because a clipped pixel is the same in every frame. The plain s.d. is 1.21× larger, because residual misregistration leaves errors along edges; on flat pixels of one brightness the two agree to 2%. The average's residual noise is σ/√N: 0.0031 of full scale (10.5 DN) at ISO 2500, 12.0 dB below a single frame, which caps a measurable denoising score near 50 dB. I treat it as the floor of every denoising score. [ISO 50 and ISO 200 values.]

### B. Demosaicing: A Mosaic With a Known Answer

Each 2 × 2 block of the *averaged* burst becomes one quarter-resolution pixel (R site, mean of the G sites, B site), which is re-mosaicked and demosaicked. Homework 5 showed why the average matters: against a single-frame reference, MHC scored 0.60 dB *below* bilinear on my checkerboard, because its filters pass about twice bilinear's noise (Σw² of 0.56 against 0.25 for green), and box-averaging the reference 2×, which halves its noise, turned that into a 0.62 dB lead. The construction has an error of its own that averaging doesn't remove: the R, G and B values of one quarter-resolution pixel come from sites up to a pixel apart, and together with the lens's lateral chromatic aberration this puts a colored fringe on sharp edges of the reference itself.

## Decisions and Verification Statement

**What I decided, and what I rejected.**

* **The demosaicing reference comes from the averaged burst, not a single frame.** Against a single frame MHC lost to bilinear by 0.60 dB on my checkerboard; on a reference with half the noise it won by 0.62 dB (Homework 5).
* **Shift estimates on a DFT-friendly crop.** On the full 2016-wide green plane, `cv2.phaseCorrelate` reports a 0.5 px shift for an image against itself.
* **The noise model uses the analog gain, not the ISO in the file.** Above about ISO 430 this phone records the extra ISO only as `BaselineExposure`; an ISO 5000 burst gave the same raw data as ISO 2500 (Homework 7).
* **Alignment with `cv2.phaseCorrelate`, not scikit-image.** On known shifts OpenCV's error was at most 0.11 px, scikit-image's 0.17 px.
